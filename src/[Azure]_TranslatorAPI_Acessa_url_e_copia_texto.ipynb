```python
import requests
from bs4 import BeautifulSoup

def extract_text_from_url(url):
    response = requests.get(url)
    try:
      if response.status_code != 200:
          raise ValueError(f"Failed to fetch URL: {url}")
      else:
        soup = BeautifulSoup(response.text, 'html.parser')
        print(f"Successfully fetched URL: {url}")
        #Particionar texto
        for script_or_style in soup(["script", "style"]):
          script_or_style.decompose()
          texto = soup.get_text(separator=" ")
          #limpar texto
          lines = (line.strip() for line in texto.splitlines())
          chunks = (phrase.strip() for line in lines for phrase in line.split("  "))
          texto_limpo = '\n'.join(chunk for chunk in chunks if chunk)
          return texto_limpo

    except Exception as e:
        print(f"Error: {e}")
        return None

extract_text_from_url("https://dev.to/satyam_gupta_0d1ff2152dcc/python-tuples-the-ultimate-guide-to-immutable-sequences-4lpf")


```

```python
from langchain_openai import AzureChatOpenAI

client = AzureChatOpenAI(azure_endpoint="https://openai-brsth-dev-yuji-001.openai.azure.com/",
                         api_key="<API_KEY>",
                         api_version="2024-10-21",
                         deployment_name="gpt-4o-mini",
                         max_retries=3
                        )

def translate_article(txt, lang):
  messages = [
      {"role": "system", "content": "Você atua como um tradutor de textos"},
      {"role": "user", "content": f"traduza o seguinte texto {txt} para o idioma {lang} e responda em markdown"}
  ]

  response = client.invoke(messages)
  print(response.content) # Corrected to access content attribute
  return response.content

translate_article("Python Tuples: The Ultimate Guide to Immutable Sequences", "português")
```

```python
url = "https://dev.to/satyam_gupta_0d1ff2152dcc/python-tuples-the-ultimate-guide-to-immutable-sequences-4lpf"

def tradutor_de_url(url):
  text = extract_text_from_url(url)
  article = translate_article(text, "pt-br")
  return article

print(tradutor_de_url(url))
```
